# 🧠 MCP Server

code: https://github.com/deep-div/MCP

This repository contains a sample **Multi-Component Protocol (MCP)** server implementation using `FastMCP`. It connects LLMs (like Gemini or OpenAI) to custom tools and provides agentic workflows.

## 📦 Features

- 🛠️ Build custom tools for LLMs  
- 🧠 Serve multiple agents with isolated toolsets  
- ⚡ FastAPI-based MCP server using `FastMCP`  
- 🔌 Easily connect with MCP clients
